```python
import cv2 as cv
import dlib
import os
import glob
import shutil
from tqdm import tqdm
```

```python
# 实例化面部检测对象
detector =dlib.get_frontal_face_detector()
```

```python
def removeUselessData(root,failedRoot):
    files = glob.glob(os.path.join(root,'*.jpg'))
    
    # tqdm 遍历面部图像
    for i in tqdm(range(len(files))):
        imgFile = files[i]
        frame = cv.imread(imgFile)
        # BGR 三通道的图像转换为 RGB 三通道
        img = cv.cvtColor(frame,cv.COLOR_BGR2RGB)
        dets = detector(img,1) #  unsample=1 防误判
        if(len(dets) == 0):
            class_name,ext = os.path.splitext(os.path.basename(imgFile))
            shutil.move(imgFile,failedRoot + '/' + class_name + '.jpg')
```

```python
removeUselessData('./data/OpenFace/' , './data/OpenFailed/')
```

```python
removeUselessData('./data/ClosedFace//' , './data/CloseFailed/')
```
